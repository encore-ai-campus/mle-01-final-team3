# 건강기능식품 원료별 정보 게시판 수집

식품안전나라 게시판의 영양성분, 기능성 원료, 개별인정원료 목록과 첨부 보고서를 수집한다.

- 영양성분: 게시판 분류번호 `1205`, 내부 식별자 `NUTRIENT:1-x`
- 기능성 원료: 게시판 분류번호 `1206`, 내부 식별자 `NOTIFIED:2-x`
- 개별인정원료: 게시판 분류번호 `1207`, 내부 식별자 `INDIVIDUAL:인정번호`

게시판 목록은 화면 HTML이 아니라 `boardList.do` JSON 응답으로 수집하고, 첨부파일 메타데이터도 같은 응답에서 가져온다. 기존 API CSV는 교차검증에만 사용한다.

## 1. 기존 API 인정번호 정규화

```python
import re
import pandas as pd


def normalize_recognition_no(value: object) -> str | None:
    if pd.isna(value):
        return None

    text = str(value).strip()
    is_new = "new" in text.lower()

    match = re.search(r"(\d{4})\s*-\s*(\d+)", text)
    if not match:
        return None

    normalized = f"{match.group(1)}-{int(match.group(2))}"
    return f"NEW-{normalized}" if is_new else normalized
```

```python
api = pd.read_csv(
    "../data/건강기능식품_기능성_원료인정현황.csv",
    encoding="utf-8-sig",
    dtype=str,
)

api["recognition_no"] = api[
    "HF_FNCLTY_MTRAL_RCOGN_NO"
].map(normalize_recognition_no)
```

## 2. 게시판 수집 설정

```python
import hashlib
import json
import math
import time
import unicodedata
from datetime import datetime
from pathlib import Path
from zoneinfo import ZoneInfo

import requests
from pypdf import PdfReader

BOARD_HOME_URL = (
    "https://www.foodsafetykorea.go.kr/portal/board/board.do"
    "?menu_grp=MENU_NEW01&menu_no=2660"
)
BOARD_LIST_URL = "https://www.foodsafetykorea.go.kr/portal/board/boardList.do"
FILE_DOWNLOAD_URL = (
    "https://www.foodsafetykorea.go.kr/common/downloadAttchdFile.do"
)
DETAIL_URL_TEMPLATE = (
    "https://www.foodsafetykorea.go.kr/portal/board/boardDetail.do"
    "?bbs_no=bbs987&menu_grp=MENU_NEW01&menu_no=2660&ntctxt_no={ntctxt_no}"
)

CATEGORY_CONFIG = {
    "영양성분": {"ctgry_no": "1205", "idx_1": "0", "prefix": "NUTRIENT"},
    "기능성 원료": {"ctgry_no": "1206", "idx_1": "1", "prefix": "NOTIFIED"},
    "개별인정원료": {"ctgry_no": "1207", "idx_1": "2", "prefix": "INDIVIDUAL"},
}

PAGE_SIZE = 100
REQUEST_INTERVAL = 0.5
DOWNLOAD_REPORTS = True
EXTRACT_PDF_TEXT = True

DATA_DIR = Path("../data")
BOARD_DATA_DIR = DATA_DIR / "건강기능식품_원료별정보"
REPORT_DIR = BOARD_DATA_DIR / "reports"
BOARD_DATA_DIR.mkdir(parents=True, exist_ok=True)
REPORT_DIR.mkdir(parents=True, exist_ok=True)

session = requests.Session()
session.headers.update(
    {
        "User-Agent": "Mozilla/5.0 (compatible; MLE-Team3-Research/1.0)",
        "Referer": BOARD_HOME_URL,
    }
)
```

## 3. 제목 파싱 및 내부 식별자 생성

고시형 항목은 `1-27`, `2-69` 형식의 번호를, 개별인정원료는 제목 속 인정번호를 식별자로 사용한다. 파일이 없는 최신 게시물도 목록에서 빠지지 않도록 한 행을 유지한다.

```python
def parse_board_title(category: str, title: str) -> dict:
    title = str(title).strip()
    source_code = None
    recognition_no = None
    ingredient_name = title

    if category in {"영양성분", "기능성 원료"}:
        match = re.match(r"^(\d+)\s*-\s*(\d+)\s+(.+?)\s*$", title)
        if match:
            source_code = f"{int(match.group(1))}-{int(match.group(2))}"
            ingredient_name = match.group(3).strip()
    elif category == "개별인정원료":
        recognition_no = normalize_recognition_no(title)
        suffix = re.search(
            r"\([^()]*(?:New\s*)?제?\s*\d{4}\s*-\s*\d+\s*호?\)\s*$",
            title,
            flags=re.IGNORECASE,
        )
        if suffix:
            ingredient_name = title[: suffix.start()].strip()

    return {
        "source_code": source_code,
        "recognition_no": recognition_no,
        "ingredient_name": ingredient_name,
    }


def parse_company_name(category: str, title: str) -> str | None:
    if category != "개별인정원료":
        return None
    match = re.search(
        r"\((?P<company>.*?),\s*(?:New\s*)?제?\s*\d{4}\s*-\s*\d+\s*호?\)\s*$",
        str(title).strip(),
        flags=re.IGNORECASE,
    )
    return match.group("company").strip() if match else None


def build_source_id(
    category: str, source_code: str | None, recognition_no: str | None
) -> str | None:
    prefix = CATEGORY_CONFIG[category]["prefix"]
    key = recognition_no if category == "개별인정원료" else source_code
    return f"{prefix}:{key}" if key else None


def flatten_board_rows(rows: list[dict], collected_at: str) -> list[dict]:
    flattened = []
    for row in rows:
        category = row.get("ctgrynm")
        title = row.get("titl", "")
        parsed = parse_board_title(category, title)
        common = {
            "source_id": build_source_id(category, parsed["source_code"], parsed["recognition_no"]),
            "category": category,
            "board_order_no": row.get("no"),
            "ntctxt_no": row.get("ntctxt_no"),
            "title": title,
            **parsed,
            "company_name_from_title": parse_company_name(category, title),
            "author": row.get("cretr_nm"),
            "registered_at": row.get("cret_dtm"),
            "last_updated_at": row.get("last_updt_dtm"),
            "view_count": row.get("inqry_cnt"),
            "detail_url": DETAIL_URL_TEMPLATE.format(ntctxt_no=row.get("ntctxt_no")),
            "collected_at": collected_at,
        }
        files = row.get("fileList") or [None]
        for file_info in files:
            file_info = file_info or {}
            flattened.append(
                {
                    **common,
                    "report_attachment_no": file_info.get("atch_file_no"),
                    "report_file_type": file_info.get("file_type_cd"),
                    "report_file_path": file_info.get("file_path"),
                    "report_filename": file_info.get("logic_file_nm"),
                    "report_physical_filename": file_info.get("physic_file_nm"),
                    "report_file_size": file_info.get("file_mg"),
                    "report_ecm_file_no": file_info.get("ecm_file_no"),
                    "report_download_endpoint": FILE_DOWNLOAD_URL if file_info else None,
                }
            )
    return flattened
```

## 4. 세 범주 목록 수집

`start_idx`는 행 번호가 아니라 1부터 시작하는 페이지 번호다. 첫 응답의 `total_cnt`로 마지막 페이지를 계산하고, 각 범주를 끝까지 순회한다. 원본 JSON도 함께 저장해 파싱 규칙 변경 시 다시 활용할 수 있게 한다.

```python
def build_board_payload(category: str, page_no: int, page_size: int) -> dict:
    config = CATEGORY_CONFIG[category]
    return {
        "menu_no": "2660",
        "menu_grp": "MENU_NEW01",
        "bbs_no": "bbs987",
        "ctgry_no": config["ctgry_no"],
        "start_idx": str(page_no),
        "nticmatr_yn": "N",
        "bbs_type_cd": "01",
        "ans_yn": "N",
        "idx_1": config["idx_1"],
        "order_type": "01",
        "ctgry_type_cd": "CTG_TYPE01",
        "list_img_use_yn": "N",
        "atch_file_posbl_yn": "Y",
        "cmt_yn": "N",
        "kword_use_yn": "N",
        "natn_cd_use_yn": "N",
        "tag_use_yn": "N",
        "meta_use_yn": "N",
        "search_type": "title",
        "search_keyword": "",
        "show_cnt": str(page_size),
    }


def fetch_board_page(
    session: requests.Session, category: str, page_no: int, page_size: int
) -> dict:
    response = session.post(
        BOARD_LIST_URL,
        data=build_board_payload(category, page_no, page_size),
        timeout=30,
    )
    response.raise_for_status()
    data = response.json()
    if "list" not in data or "total_cnt" not in data:
        raise ValueError(f"예상하지 못한 게시판 응답: {data}")
    return data


def collect_board_category(
    session: requests.Session,
    category: str,
    page_size: int = PAGE_SIZE,
    request_interval: float = REQUEST_INTERVAL,
) -> tuple[list[dict], dict]:
    first = fetch_board_page(session, category, page_no=1, page_size=page_size)
    total_count = int(first["total_cnt"])
    page_count = max(1, math.ceil(total_count / page_size))
    pages = [first]

    for page_no in range(2, page_count + 1):
        time.sleep(request_interval)
        pages.append(fetch_board_page(session, category, page_no, page_size))

    rows = [row for page in pages for row in page["list"]]
    if len(rows) != total_count:
        raise RuntimeError(
            f"{category}: 응답 건수 {len(rows):,} != total_cnt {total_count:,}"
        )
    raw = {"category": category, "total_count": total_count, "pages": pages}
    return rows, raw


collected_at = datetime.now(ZoneInfo("Asia/Seoul")).isoformat(timespec="seconds")
all_rows = []
raw_collections = {}

for category in CATEGORY_CONFIG:
    category_rows, raw = collect_board_category(session, category)
    all_rows.extend(category_rows)
    raw_collections[category] = raw
    print(f"{category}: {len(category_rows):,}건 수집")
    time.sleep(REQUEST_INTERVAL)

board_df = pd.DataFrame(flatten_board_rows(all_rows, collected_at))
board_df["registered_at"] = pd.to_datetime(
    board_df["registered_at"], errors="coerce"
)
board_df = board_df.sort_values(
    ["category", "registered_at", "ntctxt_no"], ascending=[True, False, False]
).reset_index(drop=True)

list_path = BOARD_DATA_DIR / "건강기능식품_원료별정보_목록.csv"
raw_path = BOARD_DATA_DIR / "건강기능식품_원료별정보_raw.json"
board_df.to_csv(list_path, index=False, encoding="utf-8-sig")
raw_path.write_text(
    json.dumps(raw_collections, ensure_ascii=False, indent=2), encoding="utf-8"
)
print(f"목록 저장: {list_path} ({len(board_df):,}행)")
display(board_df.head())
```

## 5. 기존 API와 교차검증

개별인정원료는 인정번호를 기준으로 API 수록 여부를 판정한다. 기능성 원료와 영양성분은 API의 제품명과 원료명을 정규화해 참고용으로 비교한다. `BOARD_ONLY`는 최신 게시물이거나 API 필드 차이일 수 있으므로 원문을 우선 확인한다.

```python
def normalize_name(value: object) -> str | None:
    if pd.isna(value):
        return None
    text = unicodedata.normalize("NFKC", str(value)).lower()
    normalized = re.sub(r"[^0-9a-z가-힣]+", "", text)
    return normalized or None


unique_board = board_df.drop_duplicates("source_id").copy()
unique_board["normalized_ingredient_name"] = unique_board["ingredient_name"].map(
    normalize_name
)
unique_board["api_match_status"] = "NOT_CHECKED"

recognition_api_path = DATA_DIR / "건강기능식품_기능성_원료인정현황.csv"
if recognition_api_path.exists():
    recognition_api = pd.read_csv(recognition_api_path, encoding="utf-8-sig", dtype=str)
    recognition_api["recognition_no"] = recognition_api[
        "HF_FNCLTY_MTRAL_RCOGN_NO"
    ].map(normalize_recognition_no)
    recognition_lookup = (
        recognition_api[["recognition_no", "APLC_RAWMTRL_NM", "BSSH_NM", "PRMS_DT"]]
        .dropna(subset=["recognition_no"])
        .drop_duplicates("recognition_no", keep="last")
        .rename(
            columns={
                "APLC_RAWMTRL_NM": "api_ingredient_name",
                "BSSH_NM": "api_company_name",
                "PRMS_DT": "api_registered_at",
            }
        )
    )
    unique_board = unique_board.merge(recognition_lookup, on="recognition_no", how="left")
    mask = unique_board["category"].eq("개별인정원료")
    unique_board.loc[mask, "api_match_status"] = unique_board.loc[
        mask, "api_ingredient_name"
    ].map(lambda value: "MATCHED" if pd.notna(value) else "BOARD_ONLY")
    unique_board.loc[mask, "ingredient_name_match"] = (
        unique_board.loc[mask, "ingredient_name"].map(normalize_name)
        == unique_board.loc[mask, "api_ingredient_name"].map(normalize_name)
    )
    unique_board.loc[mask, "company_name_match"] = (
        unique_board.loc[mask, "company_name_from_title"].map(normalize_name)
        == unique_board.loc[mask, "api_company_name"].map(normalize_name)
    )
    board_dates = pd.to_datetime(
        unique_board.loc[mask, "registered_at"], errors="coerce"
    ).dt.date
    api_dates = pd.to_datetime(
        unique_board.loc[mask, "api_registered_at"], errors="coerce"
    ).dt.date
    unique_board.loc[mask, "registered_date_match"] = board_dates == api_dates

item_api_path = DATA_DIR / "건강기능식품_품목제조신고.csv"
if item_api_path.exists():
    item_api = pd.read_csv(item_api_path, encoding="utf-8-sig", dtype=str)
    api_names = set(item_api.get("PRDCT_NM", pd.Series(dtype=str)).map(normalize_name).dropna())
    mask = unique_board["category"].isin(["영양성분", "기능성 원료"])
    unique_board.loc[mask, "api_match_status"] = unique_board.loc[
        mask, "normalized_ingredient_name"
    ].map(lambda value: "MATCHED_BY_NAME" if value in api_names else "BOARD_ONLY")

comparison_path = BOARD_DATA_DIR / "건강기능식품_원료별정보_API비교.csv"
unique_board.to_csv(comparison_path, index=False, encoding="utf-8-sig")
print(unique_board.groupby(["category", "api_match_status"]).size())
display(unique_board.query("api_match_status == 'BOARD_ONLY'").head(20))
```

## 6. 심사결과보고서 PDF/HWP 다운로드 (선택)

목록 검토 후 설정 셀의 `DOWNLOAD_REPORTS = True`로 변경해 실행한다. 게시판 JSON에 들어 있는 실제 파일 경로와 파일명을 다운로드 요청에 사용하며, 같은 첨부번호는 한 번만 받는다. 무리한 병렬 다운로드는 하지 않는다.

```python
def safe_filename(value: str, max_length: int = 180) -> str:
    cleaned = re.sub(r'[<>:"/\\|?*]+', "_", str(value)).strip(" .")
    return cleaned[:max_length] or "unnamed"


def download_report(
    session: requests.Session, row: pd.Series, report_root: Path, overwrite: bool = False
) -> dict:
    category_dir = report_root / safe_filename(row["category"])
    category_dir.mkdir(parents=True, exist_ok=True)
    filename = safe_filename(row["report_filename"])
    target = category_dir / f"{row['ntctxt_no']}_{filename}"

    if target.exists() and not overwrite:
        content = target.read_bytes()
        return {"status": "SKIPPED_EXISTS", "local_path": str(target), "sha256": hashlib.sha256(content).hexdigest()}

    payload = {
        "menu_no": "2660",
        "menu_grp": "MENU_NEW01",
        "bbs_no": "bbs987",
        "filePath": row["report_file_path"],
        "fileName": row["report_physical_filename"],
        "orgFileName": row["report_filename"],
        "file_type_cd": row["report_file_type"],
        "ecm_file_no": row["report_ecm_file_no"],
    }
    response = session.post(FILE_DOWNLOAD_URL, data=payload, timeout=120)
    response.raise_for_status()
    content = response.content
    if not content:
        raise ValueError(f"빈 첨부파일 응답: {row['report_filename']}")
    target.write_bytes(content)
    return {"status": "DOWNLOADED", "local_path": str(target), "sha256": hashlib.sha256(content).hexdigest()}


download_manifest = []
if DOWNLOAD_REPORTS:
    attachments = (
        board_df.dropna(subset=["report_attachment_no"])
        .drop_duplicates("report_attachment_no")
    )
    for _, attachment in attachments.iterrows():
        try:
            result = download_report(session, attachment, REPORT_DIR)
        except Exception as exc:
            result = {"status": "ERROR", "local_path": None, "sha256": None, "error": repr(exc)}
        download_manifest.append(
            {"report_attachment_no": attachment["report_attachment_no"], **result}
        )
        time.sleep(REQUEST_INTERVAL)

    download_df = pd.DataFrame(download_manifest)
    manifest_path = BOARD_DATA_DIR / "심사결과보고서_다운로드_목록.csv"
    download_df.to_csv(manifest_path, index=False, encoding="utf-8-sig")
    print(download_df["status"].value_counts(dropna=False))
else:
    print("첨부파일 다운로드 생략: DOWNLOAD_REPORTS = True로 바꾸면 실행됩니다.")
```

## 7. PDF 본문 텍스트 추출 (선택)

`EXTRACT_PDF_TEXT = True`일 때 다운로드된 PDF를 페이지 단위 텍스트로 저장한다. 스캔 PDF는 별도 OCR이 필요하며, HWP/HWPX는 원본 파일과 메타데이터만 보존한다.

```python
def extract_pdf_text(pdf_path: Path) -> tuple[str, int]:
    reader = PdfReader(str(pdf_path))
    pages = []
    for page_no, page in enumerate(reader.pages, start=1):
        pages.append(f"\n\n--- page {page_no} ---\n" + (page.extract_text() or ""))
    return "".join(pages).strip(), len(reader.pages)


extraction_manifest = []
if EXTRACT_PDF_TEXT:
    if not DOWNLOAD_REPORTS or not download_manifest:
        raise RuntimeError("먼저 DOWNLOAD_REPORTS = True로 첨부파일을 다운로드하세요.")
    for item in download_manifest:
        local_path = item.get("local_path")
        if not local_path or Path(local_path).suffix.lower() != ".pdf":
            continue
        pdf_path = Path(local_path)
        try:
            text, page_count = extract_pdf_text(pdf_path)
            text_path = pdf_path.with_suffix(".txt")
            text_path.write_text(text, encoding="utf-8")
            result = {"status": "EXTRACTED", "page_count": page_count, "text_path": str(text_path)}
        except Exception as exc:
            result = {"status": "ERROR", "page_count": None, "text_path": None, "error": repr(exc)}
        extraction_manifest.append({"local_path": local_path, **result})

    extraction_df = pd.DataFrame(extraction_manifest)
    extraction_path = BOARD_DATA_DIR / "심사결과보고서_PDF텍스트_목록.csv"
    extraction_df.to_csv(extraction_path, index=False, encoding="utf-8-sig")
    print(extraction_df["status"].value_counts(dropna=False))
else:
    print("PDF 텍스트 추출 생략: EXTRACT_PDF_TEXT = True로 바꾸면 실행됩니다.")
```
